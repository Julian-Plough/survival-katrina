```python
import pandas as pd
import pyreadstat

causal, _ = pyreadstat.read_dta("D:/Dissertation(!)/Paper 3/tenure_records/causal_dataset.dta")
spatial, _ = pyreadstat.read_dta("D:/Dissertation(!)/Paper 3/tract_spatial.dta")

cols_to_drop = [c for c in causal.columns if c in spatial.columns and c != 'geopin']
causal = causal.drop(columns=cols_to_drop)
df = causal.merge(spatial, on='geopin', how='inner')
print(df.shape)
df.head()
print(f"{df['tract_2000'].notna().sum():,} / {len(df):,} rows have tract_2000")
```

```python
for name, path in [
    ("1990_census", "D:/Dissertation(!)/Paper 3/1990_census.dta"),
    ("2000_census", "D:/Dissertation(!)/Paper 3/2000_census.dta"),
    ("2010_census", "D:/Dissertation(!)/Paper 3/2010_census.dta"),
    ("2000_census_homeowner", "D:/Dissertation(!)/Paper 3/2000_census_homeowner.dta"),
    ("2010_census_homeowner", "D:/Dissertation(!)/Paper 3/2010_census_homeowner.dta"),
]:
    data, _ = pyreadstat.read_dta(path)
    peek(data, name)
```

```python
# force tract columns to int in df, handling NaNs
for col in ['tract_1990', 'tract_2000', 'tract_2010']:
    df[col] = df[col].fillna(-1).astype(int).astype('Int64').replace(-1, pd.NA)

# loop through census files, force tract key to int, merge to df
census_files = [
    ("D:/Dissertation(!)/Paper 3/1990_census.dta", 'tract_1990'),
    ("D:/Dissertation(!)/Paper 3/2000_census.dta", 'tract_2000'),
    ("D:/Dissertation(!)/Paper 3/2010_census.dta", 'tract_2010'),
    ("D:/Dissertation(!)/Paper 3/2000_census_homeowner.dta", 'tract_2000'),
    ("D:/Dissertation(!)/Paper 3/2010_census_homeowner.dta", 'tract_2010'),
]

for path, key in census_files:
    data, _ = pyreadstat.read_dta(path)
    data[key] = data[key].fillna(-1).astype(int).astype('Int64').replace(-1, pd.NA)
    df = df.merge(data, on=key, how='left')
    print(f"{path.split('/')[-1]}: {df.shape}")

df.head()
```

```python
# derive l_year and l_month from move_in (Stata date format: days since 1960-01-01)
df['move_in_date'] = pd.to_datetime(df['move_in'], unit='D', origin='1960-01-01')
df['l_year'] = df['move_in_date'].dt.year.astype('Int64')
df['l_month'] = df['move_in_date'].dt.month.astype('Int64')

# load and merge inflation file
infl, _ = pyreadstat.read_dta("D:/Dissertation(!)/Paper 3/inflation_6_21.dta")
infl['l_year'] = infl['l_year'].astype('Int64')
infl['l_month'] = infl['l_month'].astype('Int64')

df = df.merge(infl, on=['l_year', 'l_month'], how='left')
df['sale_price_adj'] = df['sale_price'] * df['inflation_6_21']
df['sale_price_adj'] = (df['sale_price'] * df['inflation_6_21']).round(-2).fillna(pd.NA).astype('Int64')

print(f"{df['sale_price_adj'].notna().sum():,} / {len(df):,} records have adjusted price")
df[['move_in', 'move_in_date', 'l_year', 'l_month', 'sale_price', 'inflation_6_21', 'sale_price_adj']].head(20)
```

```python
df = pd.read_parquet("../data/processed/assembled.parquet")
df['moved_out'] = df['move_out'].notna().astype(int)
print(df['moved_out'].value_counts())
```

```python
SCRAPE_END = 22281
df['duration'] = (df['move_out'] - df['move_in']).fillna(SCRAPE_END - df['move_in'])
df = df[df['duration'] >= 90]
print(df['duration'].describe())
print(f"{len(df):,} records after 90-day trim")
```

```python
KATRINA_DATE = 16677
df['katrina_tenure'] = (
    (df['move_in'] <= KATRINA_DATE) & 
    ((df['move_out'] >= KATRINA_DATE) | (df['current_tenure'] == 1))
).astype(int)

print(df['katrina_tenure'].value_counts())
```

```python
df['fema_category'] = 0
df.loc[df['fld_zone'].isin(['A', 'AE', 'AO', 'VE']), 'fema_category'] = 100
df.loc[(df['fld_zone'] == 'X') & (df['zone_subty'] == '0.2 PCT ANNUAL CHANCE FLOOD HAZARD'), 'fema_category'] = 500

print(df['fema_category'].value_counts())
```

```python
df['black_white_other'] = None

mask = (
    df['applicant_race_1'].notna() & 
    ~df['applicant_race_1'].isin(['', 'Missing', 
                                   'Information not provided by applicant in mail, Internet, or telephone application', 
                                   'Not applicable'])
)

df.loc[mask & ~df['applicant_race_1'].isin(['White', 'Black', 'Black or African American']), 'black_white_other'] = 0
df.loc[mask & (df['applicant_race_1'] == 'White'), 'black_white_other'] = 1
df.loc[mask & df['applicant_race_1'].isin(['Black', 'Black or African American']), 'black_white_other'] = 2

print(df['black_white_other'].value_counts(dropna=False))
```

```python
df['income_adjusted'] = pd.to_numeric(df['applicant_income'], errors='coerce') * df['inflation_6_21']

p1 = df['income_adjusted'].quantile(0.01)
p99 = df['income_adjusted'].quantile(0.99)
df['income_trim_flag'] = (~df['income_adjusted'].between(p1, p99)) & df['income_adjusted'].notna()

print(df['income_adjusted'].describe())
print(f"income trim flag: {df['income_trim_flag'].sum():,} records flagged")
```

```python
print("HMDA match status:")
print(f"  No match (empty):     {(df['applicant_race_1'] == '').sum():,}")
print(f"  Missing/not provided: {df['applicant_race_1'].isin(['Missing', 'Information not provided by applicant in mail, Internet, or telephone application']).sum():,}")
print(f"  Not applicable:       {(df['applicant_race_1'] == 'Not applicable').sum():,}")
print(f"  Usable race data:     {df['black_white_other'].notna().sum():,}")
print(f"    White:              {(df['black_white_other'] == 1).sum():,}")
print(f"    Black:              {(df['black_white_other'] == 2).sum():,}")
print(f"    Other:              {(df['black_white_other'] == 0).sum():,}")
print(f"  Total records:        {len(df):,}")
print()
print("Income status:")
print(f"  Has income data:      {df['applicant_income'].notna().sum():,}")
print(f"  Has adjusted income:  {df['income_adjusted'].notna().sum():,}")
print(f"  Income describe:")
print(df['income_adjusted'].describe())
```

```python
df['move_out_year'] = pd.to_datetime(df['move_out'], unit='D', origin='1960-01-01').dt.year
df['move_in_year'] = pd.to_datetime(df['move_in'], unit='D', origin='1960-01-01').dt.year

df['census_income'] = pd.NA
df['census_percent_black_ho'] = pd.NA

df.loc[(df['move_out_year'] >= 1985) & (df['move_out_year'] < 1995), 'census_income'] = df['income_tract_1990'] * 2.09
df.loc[(df['move_out_year'] >= 1995) & (df['move_out_year'] < 2005), 'census_income'] = df['income_tract_2000'] * 1.58
df.loc[(df['move_out_year'] >= 2005) & (df['move_out_year'] <= 2015), 'census_income'] = df['income_tract_2010'] * 1.25
df.loc[(df['move_in_year'] <= 2015) & (df['moved_out'] == 0), 'census_income'] = df['income_tract_2010'] * 1.25

df.loc[(df['move_out_year'] >= 1995) & (df['move_out_year'] < 2005), 'census_percent_black_ho'] = df['percent_black_2000_homeowner']
df.loc[(df['move_out_year'] >= 2005) & (df['move_out_year'] <= 2015), 'census_percent_black_ho'] = df['percent_black_2010_homeowner']
df.loc[(df['move_in_year'] <= 2015) & (df['moved_out'] == 0), 'census_percent_black_ho'] = df['percent_black_2010_homeowner']

df['census_income'] = pd.to_numeric(df['census_income'], errors='coerce')
df['census_percent_black_ho'] = pd.to_numeric(df['census_percent_black_ho'], errors='coerce')

print(df[['census_income', 'census_percent_black_ho']].describe())
```
